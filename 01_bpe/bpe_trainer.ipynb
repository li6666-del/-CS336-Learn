```python
import regex as re
with open("D:\CS336\CS336-Learn\data\TinyStoriesV2-GPT4-valid.txt",'r',encoding="UTF-8") as f:
    text=f.read()
```

```python
pattern = "(<\\|endoftext\\|>)"   #定义特殊标记
text_list=re.split(pattern,text)   
#print(text_list)
```

```python
PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""
```

```python
word_freq = {}#创建从字节元组到出现次数的字典
```

```python
for chunk in text_list:
    if chunk != '<|endoftext|>':
        for match in re.finditer(PAT, chunk):
            word = match.group().encode("utf-8")
            key = tuple(bytes([b]) for b in word)
            word_freq[key] = word_freq.get(key, 0) + 1
```

```python
print(word_freq)
```

```python
vocab = {}
for i in range(256):
    vocab[i] = bytes([i])
vocab[256] = '<|endoftext|>'.encode("utf-8")#初始化token字典，一开始只包括从0到255共256个字节和一个特殊标记符，建立从数字索引到token的字典
```

```python
pair_freq={}
for word,frequence in word_freq.items():
    for i in range(len(word)-1):
        key=tuple([word[i],word[i+1]])
        pair_freq[key]=pair_freq.get(key,0)+frequence
        
    
```

```python
print(pair_freq)
```

```python
the_most_pair = max(pair_freq, key=lambda x: (pair_freq[x], x))
```

```python
print(the_most_pair)
```

```python
most_together_pair=the_most_pair[0]+the_most_pair[1]
new_word_freq={}
for word,frequence in word_freq.items():
    new_word=[]
    i=0
    while i < len(word):
        if i < len(word) - 1 and word[i] == the_most_pair[0] and word[i+1] == the_most_pair[1]:
            new_word.append(most_together_pair)
            i += 2
        else:
            new_word.append(word[i])
            i += 1
    new_word = tuple(new_word)
    new_word_freq[new_word]=new_word_freq.get(new_word,0)+frequence
word_freq=new_word_freq
         
```

```python
vocab[len(vocab)] = most_together_pair#新归结出来的字节对加入token字典里面
```

```python
# 在大循环开始前
merges = []

# 每轮合并后
merges.append(the_most_pair)
```

```python
vocab_size=300
while len(vocab)<vocab_size:
    pair_freq={}
    for word,frequence in word_freq.items():
        for i in range(len(word)-1):
            key=tuple([word[i],word[i+1]])
            pair_freq[key]=pair_freq.get(key,0)+frequence
    the_most_pair = max(pair_freq, key=lambda x: (pair_freq[x], x))        
    most_together_pair=the_most_pair[0]+the_most_pair[1]
    new_word_freq={}
    for word,frequence in word_freq.items():
        new_word=[]
        i=0
        while i < len(word):
            if i < len(word) - 1 and word[i] == the_most_pair[0] and word[i+1] == the_most_pair[1]:
                new_word.append(most_together_pair)
                i += 2
            else:
                new_word.append(word[i])
                i += 1
        new_word = tuple(new_word)
        new_word_freq[new_word]=new_word_freq.get(new_word,0)+frequence
    merges.append(the_most_pair)
    vocab[len(vocab)] = most_together_pair
    word_freq=new_word_freq
```

```python
print("vocab 大小:", len(vocab))           # 应该 == 300
print("merges 数量:", len(merges))         # 应该 == 300 - 257 = 43
print("前 10 个合并:", merges[:10])
```
